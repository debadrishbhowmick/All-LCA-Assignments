```python
matrix1 = [
    [1,2,3],
    [4,5,6]
]
matrix2 = [
    [1,2,3],
    [4,5,6]
]
result = [
    [0,0,0],
    [0,0,0]
]

for i in range(len(matrix1)):
    for j in range(len(matrix2)):

        result[i][j] = matrix1[i][j] + matrix2[i][j]

print ('Matrix 1:')
for row in matrix1:
    print (row)

print ('\nMatrix 2:')
for row in matrix2:
    print (row)

print ('\n Sum of matrix 1 and matrix 2:')
for row in result:
    print (row)
```
